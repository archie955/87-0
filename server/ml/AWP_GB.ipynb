# Gradient Boosting for AWPer prediction model

```python
import pandas as pd

data = pd.read_csv("./Training_Data.csv")
data.head()
```

```python
feature_cols = [
    "firepower",
    "entrying",
    "trading",
    "opening",
    "clutching",
    "sniping",
    "utility",
    "igl",
]

x = data[feature_cols]
y = data["position"]

y.replace("Closer", 0, inplace=True)
y.replace("Linchpin", 0, inplace=True)
y.replace("Opener", 0, inplace=True)
y.replace("AWPer", 1, inplace=True)

y.head()
```

```python
import numpy as np
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    precision_score,
    recall_score,
    ConfusionMatrixDisplay,
)
from sklearn.model_selection import RandomizedSearchCV, train_test_split
from scipy.stats import randint, uniform


x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=42, stratify=y
)
```

```python
# le = LabelEncoder()
# y_train = le.fit_transform(y_train)
# y_test = le.fit_transform(y_test)
```

```python
param_dist = {
    "n_estimators": randint(100, 600),
    "max_depth": randint(3, 8),
    "learning_rate": uniform(loc=0.01, scale=0.3),
    "subsample": uniform(0.7, 0.3),
    "colsample_bytree": uniform(0.7, 0.3),
}


gb = XGBClassifier(random_state=42, n_jobs=1, objective="binary:logistic")

rand_search = RandomizedSearchCV(
    gb,
    param_distributions=param_dist,
    n_iter=5,
    cv=5,
    scoring="accuracy",
    n_jobs=2,
    random_state=42,
).fit(x_train, y_train)

best_rf = rand_search.best_estimator_

params = rand_search.best_params_

print("Best hyperparameters:", params)
```

```python
model = XGBClassifier(**params, objective="binary:logistic")
model.fit(x_train, y_train)

y_pred = model.predict(x_test)

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy: ", accuracy)
```
