Multinomial Logistic regression

```python
import pandas as pd
data = pd.read_csv("./Training_Data.csv")
data.describe()
```

```python
for col in [
    "firepower",
    "entrying",
    "trading",
    "opening",
    "clutching",
    "sniping",
    "utility",
]:
    data[col] = data[col] / 100

data.head()
```

```python
feature_cols = [
    "firepower",
    "entrying",
    "trading",
    "opening",
    "clutching",
    "sniping",
    "utility",
    "igl"
]

# train_data = data.loc[data["igl"] == 0]
train_data = data

train_data.head()
```

```python
X = train_data[feature_cols]
Y = train_data[["position", "number_of_maps"]]

Y.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.25, random_state=16
)
```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

logreg = LogisticRegression(random_state=16)

sample_weight = np.log(Y_train["number_of_maps"])

clf = logreg.fit(X_train, Y_train["position"], sample_weight)

y_pred = clf.predict(X_test)
```

```python
from sklearn.metrics import classification_report

target_names = ["AWPer", "Opener", "Closer", "Linchpin"]

print(classification_report(Y_test["position"], y_pred, target_names=target_names))
```

```python
print(clf.predict(np.array([0.7, 0.2, 0.5, 0.2, 0.6, 0.9, 0.4, 0]).reshape(1, -1)))
```

```python
print(clf.coef_, clf.intercept_)

# in order: AWPer, Closer, Linchpin, Opener
# features: ["firepower", "entrying", "trading", "opening", "clutching", "sniping", "utility"]
```

```python
print(clf.classes_)
```

```python
print(clf.feature_names_in_)
```

```python
ropz = [0.70, 0.03, 0.46, 0.49, 0.49, 0.02, 0.22, 0.0]

ropz_df = np.array(ropz).reshape(1, -1)
print(clf.predict(ropz_df))
```

```python
def closer(p):
    total = clf.intercept_[1]
    for i in range(7):
        total += p[i]*clf.coef_[1][i]
    return total

def linchpin(p):
    total = clf.intercept_[2]
    for i in range(7):
        total += p[i]*clf.coef_[2][i]
    return total

def awper(p):
    total = clf.intercept_[0]
    for i in range(7):
        total += p[i]*clf.coef_[0][i]
    return total

def opener(p):
    total = clf.intercept_[3]
    for i in range(7):
        total += p[i]*clf.coef_[3][i]
    return total
    

print(f"Ropz closer rating is {closer(ropz)}, and his linchpin rating is {linchpin(ropz)}. His awper rating is {awper(ropz)} and his opener rating is {opener(ropz)}.")
```

```python
ratings = {"awper": awper(ropz), "closer": closer(ropz), "linchpin": linchpin(ropz), "opener": opener(ropz)}
```

```python
mini = 10
min_key = "null"
for key in ratings:
    if ratings[key] < mini:
        mini = ratings[key]
        min_key = key

for key in ratings:
    ratings[key] -= mini
```

## Merging closer/linchpin together

```python
Y.replace("Linchpin", "Closer", inplace=True)
```

```python
Y.head()
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.25, random_state=16
)

logreg = LogisticRegression(random_state=16)

sample_weight = np.log(Y_train["number_of_maps"])

clf = logreg.fit(X_train, Y_train["position"], sample_weight)

y_pred = clf.predict(X_test)
```

```python
target_names = ["AWPer", "Opener", "Closer"]

print(classification_report(Y_test["position"], y_pred, target_names=target_names))
```
