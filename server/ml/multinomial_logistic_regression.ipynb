Multinomial Logistic regression

```python
import pandas as pd

col_names = [
    "name",
    "position",
    "number_of_maps",
    "firepower",
    "entrying",
    "trading",
    "opening",
    "clutching",
    "sniping",
    "utility",
]

data = pd.read_csv("./Training_Data.csv")
data.head()
```

```python
for col in [
    "firepower",
    "entrying",
    "trading",
    "opening",
    "clutching",
    "sniping",
    "utility",
]:
    data[col] = data[col] / 100

data.head()
```

```python
feature_cols = [
    "firepower",
    "entrying",
    "trading",
    "opening",
    "clutching",
    "sniping",
    "utility",
]
train_data = data.loc[data["igl"] == 0]

train_data.head()
```

```python
X = train_data[feature_cols]
Y = train_data[["position", "number_of_maps"]]

Y.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.25, random_state=16
)
```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

logreg = LogisticRegression(random_state=16)

sample_weight = np.sqrt(Y_train["number_of_maps"])

clf = logreg.fit(X_train, Y_train["position"], sample_weight)

y_pred = clf.predict(X_test)
```

```python
from sklearn.metrics import classification_report

target_names = ["AWPer", "Opener", "Closer", "Linchpin"]

print(classification_report(Y_test["position"], y_pred, target_names=target_names))
```

```python
print(logreg.predict(np.array([0.7, 0.2, 0.5, 0.2, 0.6, 0.9, 0.4]).reshape(1, -1)))
```

```python
print(clf.coef_, clf.intercept_)

# in order: AWPer, Closer, Linchpin, Opener
# features: ["firepower", "entrying", "trading", "opening", "clutching", "sniping", "utility"]
```
